### Indexing and Slicing:


```python
import numpy as np
```

```python
# Indexing:----->
a1 = np.array([1,2,3,4,5,6])
print(a1)
#Access
#Array[index]
print(a1[2])
```

```python
a1 = np.array([[1,2,3], [4,5,6]])
print(a1)
print()
print(a1[0])
print()
print(a1[1][2])
print()
print(a1[0,1])
```

```python
# Slicing:-------->>
a1 = np.array([1,2,3,4,5,6])
a1 = np.array([11,22,33,44,55,66])
print(a1)
#arr[startindex : endindex:stepvalue]
a1[1:-2:1]
a1[-2:1:-1]
```

```python
a2 = np.array([[11,22,33,44],[55,66,77,88]])
print(a2)
print()
print(a2[0][1:-1:1])
print()
print(a2[1,0:-1:1])
```

```python
arr = np.array([[11,22,33,44],[1,2,3,4],[21,32,43,54]])
print(arr)
print()
print(arr[:-1,1:-1])
```

```python
arr = np.array([[11,22,33,44],[1,2,3,4],[21,32,43,54]])
print(arr)
```

```python
print(arr[1])
print(arr[-1,2])
print(arr[1,1:])
print(arr[-1,-2::-1])
print(arr[:,2])
print(arr[1:,1])
print(arr[:2,2:])
print(arr[::-1,0])
```

Methods and Function:

```python
#Sort
a1 = np.array([1,4,5,2,3,6])
print(a1)
a2 = np.sort(a1)         # Function - new obj return
print(a2)
```

```python
#Sort
a1 = np.array([1,4,5,2,3,6])
a1.sort()
print(a1)
print(a1 [::-1])               # Method - original obj change
```

```python
#argsort
a1 = np.array([1,4,5,2,3,6])
np.argsort(a1)
```

```python
emp= np.array(['raj','pavan','ram','kunal'])
sal = np.array([22000,11000,33000,12000])
ar = np.argsort(sal)
for index in ar:
    print(emp[index])
```

```python
#Where
emp= np.array(['raj','pavan','ram','kunal'])
sal = np.array([22000,11000,33000,12000])
print(np.where(sal>20000)) # return index
sal[np.where(sal>20000)]   # return value
```

```python
#append
emp= np.array(['raj','pavan','ram','kunal'])
sal = np.array([22000,11000,33000,12000])
emp1 = np.append(emp,"vaibhav")
print(emp1)
```

```python
#insert
sal = np.array([22000,11000,33000,12000])
np.insert(sal,2,10000)
```
