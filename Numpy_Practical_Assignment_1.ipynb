Student Marks Analysis System using NumPy.

```python
# import
import numpy as np
```

```python
#Dataset : student_names and marks
students = np.array(["Amit", "Rahul", "Priya", "Sneha", "Vikas",
                     "Neha", "Rohit", "Pooja", "Akash", "Kiran"])
marks = np.array([78, 45, 92, 67, 35, 88, 56, 74, 95, 62])
```

```python
#Part 1 — Basic Array Information
print("Number of students :", len(students))
print("Dimensions         :", marks.ndim)
print("Shape              :", marks.shape)
print("Size               :", marks.size)
print("Data type          :", marks.dtype)
```

```python
#Part 2 — Sorting Operations
print("Ascending  :", np.sort(marks))               
print("Descending :", np.sort(marks)[::-1])

idx = np.argsort(marks) 
print(idx)
print("\nStudents sorted by marks (low to high):")
for n, m in zip(students[idx], marks[idx]):
    print(f"{n:<8} {m}")

print("\nTop 3 students:")
top3 = idx[::-1][:3] 
print(top3)
for n, m in zip(students[top3], marks[top3]):
    print(f"{n:<8} {m}")

print("\nLowest 3 students:")
low3 = idx[:3] 
print(low3)                                     
for n, m in zip(students[low3], marks[low3]):
    print(f"{n:<8} {m}")
```

```python
# Part 3 — Searching Operations
print("Task 1 - Marks > 70 :", students[marks > 70])
print("Task 2 - Marks < 50 :", students[marks < 50])
print("Task 3 - argwhere (marks > 80):", np.argwhere(marks > 80).flatten())
print("Task 4 - nonzero indices      :", np.nonzero(marks)[0])
print("Task 5 - count (marks > 60)   :", np.count_nonzero(marks > 60))

sorted_marks = np.sort(marks)
print(sorted_marks)
print("Task 6 - searchsorted for 70  :", np.searchsorted(sorted_marks, 70))

mask = np.isin(marks, [45, 67, 95])                 
print("Task 7 - isin [45,67,95]      :", students[mask], marks[mask])
```

```python
# Part 4 — Conditional Operations
result = np.where(marks >= 40, "Pass", "Fail")
print("Task 1 - Pass/Fail:")
print(f"{'Student':<8} {'Marks':<6} Result")
for n, m, r in zip(students, marks, result):
    print(f"{n:<9} {m:<6} {r}")

print("\nTask 2 - Passed students:", students[marks >= 40]) 
print("Task 3 - Failed students:", students[marks < 40])

# Task 4 - np.select - grade
conditions = [marks < 40,
              (marks >= 40) & (marks <= 59),
              (marks >= 60) & (marks <= 74),
              (marks >= 75) & (marks <= 89),
              (marks >= 90) & (marks <= 100)]
grades = np.select(conditions, ["Fail", "C", "B", "A", "A+"], default="Invalid")
print("\nTask 4 - Grades:")
print(f"{'Student':<8} {'Marks':<6} Grade")
for n, m, g in zip(students, marks, grades):
    print(f"{n:<8} {m:<6} {g}")

# Task 5 - invalid marks clip karne (0 te 100)
invalid = np.array([78, 45, 110, 67, -5, 88])
print("\nTask 5 - Original :", invalid)
print("         Clipped  :", np.clip(invalid, 0, 100))

# Task 6 - Logical AND (60 <= marks <= 80)
and_mask = np.logical_and(marks >= 60, marks <= 80)
print("\nTask 6 - marks 60 to 80:", students[and_mask], marks[and_mask])

# Task 7 - Logical OR (marks < 40 ki > 90)
or_mask = np.logical_or(marks < 40, marks > 90)
print("Task 7 - marks <40 or >90:", students[or_mask], marks[or_mask])
```

```python
# Part 5 — Statistical Analysis
print("Total marks    :", np.sum(marks))
print("Average        :", np.mean(marks))
print("Median         :", np.median(marks))
print("Minimum        :", np.min(marks))
print("Maximum        :", np.max(marks))
print("Std deviation  :", np.std(marks))
print("Variance       :", np.var(marks))
print("25th percentile:", np.percentile(marks, 25))
print("50th percentile:", np.percentile(marks, 50))
print("75th percentile:", np.percentile(marks, 75))
print("Index of highest:", np.argmax(marks))
print("Index of lowest :", np.argmin(marks))
print("Highest student :", students[np.argmax(marks)])
print("Lowest student  :", students[np.argmin(marks)])
```

```python
# Part 6 — Missing Value Handling
m_nan = np.array([78, 45, np.nan, 67, 35, np.nan, 56, 74, 95, 62])
print("Array with NaN :", m_nan)
print("isnan          :", np.isnan(m_nan))
print("NaN count      :", np.isnan(m_nan).sum())
print("nanmean        :", np.nanmean(m_nan))
print("nanmedian      :", np.nanmedian(m_nan))
print("nansum         :", np.nansum(m_nan))
print("nanmin         :", np.nanmin(m_nan))
print("nanmax         :", np.nanmax(m_nan))
print("nanstd         :", np.nanstd(m_nan))

filled = np.where(np.isnan(m_nan), np.nanmean(m_nan), m_nan)    
print("NaN replaced with mean:", filled)
```

```python
# Part 7: Infinity Handling 
inf_arr = np.array([10, 20, np.nan, np.inf, -np.inf, 50])
print("Array    :", inf_arr)
print("isinf    :", np.isinf(inf_arr))
print("isfinite :", np.isfinite(inf_arr))
```

```python
# Part 8: Random Number Generation  
rng = np.random.default_rng(42)
 
print("Task 1 - Random marks     :", rng.integers(1, 101, size=10))
print("Task 2 - Random floats    :", rng.random(5))
 
courses = np.array(["Python", "Java", "Data Science", "Data Analytics"])
print("Task 3 - Random courses   :", rng.choice(courses, 5))   # repeat challa
print("Task 4 - Random students  :", rng.choice(students, 3, replace=False))
print("Task 5 - Normal (60,10)   :", rng.normal(60, 10, 10))
print("Task 6 - Uniform (10,20)  :", rng.uniform(10, 20, 5))
print("Task 7 - Binomial (10,0.5):", rng.binomial(10, 0.5, 10))
print("Task 8 - Poisson (lam=3)  :", rng.poisson(3, 10))
 
shuffled = students.copy()          # original la dhakka nako mhanun copy
rng.shuffle(shuffled)
print("Task 9 - Shuffled copy    :", shuffled)
print("         Original intact  :", students)
 
print("Task 10 - Original :", students)
print("          Permuted :", rng.permutation(students))
```

```python
# Part 9 - File Handling
np.save("student_marks.npy", marks)          # Task 1
print("Task 2 - Loaded npy:", np.load("student_marks.npy"))
 
np.savez("student_data.npz", students=students, marks=marks)   # Task 3
data = np.load("student_data.npz")                             # Task 4
print("Task 4 - Names :", data["students"])
print("         Marks :", data["marks"])
 
np.savetxt("marks.csv", marks, delimiter=",", fmt="%d")   # Task 5
print("Task 6 - Loaded CSV:", np.loadtxt("marks.csv", delimiter=","))
 
arr2d = np.array([[78, 45, 92],
                  [67, 35, 88],
                  [56, 74, 95]])                          # Task 7
np.savetxt("student_subject_marks.csv", arr2d, delimiter=",", fmt="%d")
print("Task 7 - 2D CSV loaded:")
print(np.loadtxt("student_subject_marks.csv", delimiter=","))
```

```python
# Final Task - Student Report
# =========================================================
print("\n===== Student Report =====")
print(f"{'Student':<10}{'Marks':<8}Grade")
print("-" * 27)
for n, m, g in zip(students, marks, grades):
    print(f"{n:<10}{m:<8}{g}")
print("-" * 27)

print("Total Students  :", students.size)
print("Total Marks     :", marks.sum())
print("Average Marks   :", marks.mean())
print("Highest Marks   :", marks.max())
print("Lowest Marks    :", marks.min())
print("Passed Students :", np.count_nonzero(marks >= 40))
print("Failed Students :", np.count_nonzero(marks < 40))
```
