```python
# ---------- 0-D Array (Scalar) ----------
ar0 = numpy.array(78)      # stores a single value, has no dimension
print(ar0) 
print(ar0.ndim)            # ndim checks the dimension -> Output: 0 (0-D array)
```

```python
# ---------- 1-D Array (Vector) ----------
ar1 = numpy.array([10,20,30,40])   # multiple values in a single row (list)
print(ar1)   
print(ar1.ndim)                       
# ar1.ndim would give output 1 (1-D array)
```

```python
# ---------- 2-D Array (Matrix) ----------
ar2 = numpy.array([[11,22],[33,44],[55,66]])   # rows and columns (list of lists)
print(ar2)
print(ar2.ndim)
# Output:
# [[11 22]
#  [33 44]
#  [55 66]]
# ar2.ndim would give output 2 (2-D array)
```

```python
# ---------- 3-D Array ----------
ar3 = numpy.array([
                    [[1,2,3,4], [5,6,7,8], [11,22,33,44]],
                    [[12,13,14,15], [21,32,55,66], [1,2,3,4]]
                  ])

print(ar3)
print(ar3.ndim)
print(ar3.shape)
```

```python
# Array Properties (ndim, shape, size, dtype):------>>

arr = numpy.array([[11, 22, 33], [44, 55, 66]])

print(arr)              # prints the array itself

print(arr.ndim)         # checks number of dimensions -> Output: 2

print(arr.shape)        # checks structure (rows, columns) -> Output: (2, 3)

print(arr.size)         # checks total number of elements -> Output: 6

print(arr.dtype)        # checks data type of elements -> Output: int64

print(type(arr))        # confirms it is a numpy ndarray object
```
